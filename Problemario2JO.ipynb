### Marco Teórico
En esta sección, se explica el funcionamiento básico de los siguientes componentes:

- Transformador: Dispositivo electromagnético que convierte voltaje de un nivel a otro, en este caso de 120V AC a 12V AC.
Puente de diodos: Dispositivo que rectifica la señal de AC a DC, utilizando cuatro diodos para permitir solo el paso de la corriente en una dirección.
- Circuito RC: Filtro utilizado para reducir las variaciones o rizado en la señal rectificada.
- Regulador 7805: Dispositivo que toma una entrada de 12V DC y la reduce a un nivel de 5V DC constante para alimentar circuitos electrónicos sensibles como el Arduino Uno.
- Arduino Uno y NFC: Microcontrolador que lee tarjetas NFC y guarda los registros de entrada y salida.

### Diseño del Circuito
1. Descripción del Transformador
Se utiliza un transformador que reduce el voltaje de 120V AC a 12V AC. Este dispositivo se selecciona con base en la capacidad de corriente y el voltaje de salida necesarios para alimentar el resto del circuito.

2. Rectificación con Puente de Diodos
El puente de diodos es un conjunto de cuatro diodos conectados en una configuración que convierte la señal de 12V AC a 12V DC aproximadamente, aunque con cierto rizado.

3. Filtro RC
Para suavizar la señal de salida del puente de diodos, se utiliza un filtro RC. Este circuito consta de un resistor y un capacitor conectados en serie, donde el capacitor almacena energía y reduce el rizado. El diseño del filtro depende de la frecuencia de la señal rectificada y de los componentes seleccionados.

4. Regulador 7805
El regulador de voltaje 7805 convierte los 12V DC a 5V DC, necesarios para alimentar el Arduino. Este componente es capaz de suministrar hasta 1.5A de corriente, lo que es suficiente para alimentar al microcontrolador y el lector NFC.

5. Arduino Uno y Lector NFC
El Arduino Uno se encarga de leer los datos de las tarjetas NFC y registrar la hora de entrada y salida. Este sistema permitirá llevar un control de acceso mediante la lectura de las identificaciones de los usuarios.

### Ecuación Diferencial del Circuito RC

El circuito RC está compuesto por un resistor \( R \) y un capacitor \( C \) en serie. La ecuación diferencial que describe el comportamiento del circuito es:

\begin{equation}
V_{in}(t) = R \cdot C \cdot \frac{dV_C(t)}{dt} + V_C(t)
\end{equation}

donde:

  V_{in}(t) es la señal de voltaje de entrada (después del puente de diodos),
  
  V_C(t) es el voltaje en el capacitor (que será la salida suavizada),

  R es la resistencia del circuito,

  C  es la capacitancia del circuito.


### Aplicación de la Transformada de Laplace

Aplicamos la transformada de Laplace a ambos lados de la ecuación diferencial. Usamos la propiedad de la transformada de Laplace para la derivada:

\begin{equation}
\mathcal{L}\left\{ \frac{dV_C(t)}{dt} \right\} = s V_C(s) - V_C(0)
\end{equation}

Asumiendo que el capacitor está inicialmente descargado, es decir, \( V_C(0) = 0 \), la ecuación diferencial en el dominio de Laplace se convierte en:

\begin{equation}
V_{in}(s) = R \cdot C \cdot s \cdot V_C(s) + V_C(s)
\end{equation}

Factorizando \( V_C(s) \):

\begin{equation}
V_{in}(s) = V_C(s) \cdot (R \cdot C \cdot s + 1)
\end{equation}

Por lo tanto, la ecuación en términos de \( V_C(s) \) es:

\begin{equation}
V_C(s) = \frac{V_{in}(s)}{R \cdot C \cdot s + 1}
\end{equation}

### Análisis en el Dominio de la Frecuencia

La función de transferencia \( H(s) \) es:

\begin{equation}
H(s) = \frac{1}{R \cdot C \cdot s + 1}
\end{equation}

Esta ecuación muestra cómo el circuito RC atenúa diferentes frecuencias. Cuanto mayor sea la frecuencia de la señal de entrada, menor será la respuesta del circuito, lo que significa que las componentes de alta frecuencia (rizado) se suavizan.

### Transformada Inversa de Laplace

Para obtener la respuesta en el dominio del tiempo, aplicamos la transformada inversa de Laplace a la función de transferencia estándar para un circuito RC:

\begin{equation}
\mathcal{L}^{-1} \left\{ \frac{1}{R \cdot C \cdot s + 1} \right\} = e^{-\frac{t}{R C}}
\end{equation}

Esta ecuación representa la respuesta natural del circuito en el tiempo. Si la entrada \( V_{in}(t) \) es una señal constante o una señal con rizado, la salida \( V_{out}(t) \) será la versión suavizada de la misma tras pasar por el filtro RC.

### Conclusíon:

El análisis mediante la transformada de Laplace permite comprender cómo el circuito RC filtra las frecuencias de la señal de entrada, suavizando el rizado y permitiendo obtener una señal de corriente continua más estable para alimentar componentes sensibles como un Arduino Uno. El uso de la transformada inversa de Laplace nos proporciona la respuesta en el tiempo del sistema, mostrando el comportamiento real del voltaje en el capacitor.
