# Transformada inversa de Laplace

1.
$$
\mathcal{L}^{-1} \left\{ \frac{1}{s^3 + 5s} \right\}
$$

```python
ruta_imagen = '/content/drive/MyDrive/Problemario2MAC/Inversas/1.jpg'
imagen_redimensionada = Image.open(ruta_imagen).resize((900, 900))
display(imagen_redimensionada)
```

```python
import sympy as sp

# Definir la variable s y t
s, t = sp.symbols('s t')

# Función de Laplace
F_s = 1 / (s**3 + 5*s)

# Transformada inversa de Laplace
f_t = sp.inverse_laplace_transform(F_s, s, t)
print(f"Transformada inversa de 1/(s^3 + 5s): {f_t}")
```
